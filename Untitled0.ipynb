```python
import pandas as pd
import matplotlib.pyplot as plt

print("Loading local NOAA CO2 data...")
df_co2 = pd.read_csv('/co2_mm_mlo.csv', comment='#')

df_co2 = df_co2[df_co2['average'] > 0]
df_co2['date'] = pd.to_datetime(df_co2[['year', 'month']].assign(day=1))

plt.figure(figsize=(10, 5))
plt.plot(df_co2['date'], df_co2['average'], color='#d62728', linewidth=1.5)
plt.title('Atmospheric CO2 Concentration (1958 - 2026)', fontsize=14)
plt.xlabel('Year', fontsize=12)
plt.ylabel('CO2 (ppm)', fontsize=12)
plt.grid(True, linestyle='--', alpha=0.7)
plt.tight_layout()
plt.show()
```

```python
import pandas as pd
import matplotlib.pyplot as plt
import os

# Handle Colab upload location variations automatically
file_path = '/GlobalLandTemperaturesByCity.csv' if os.path.exists('/GlobalLandTemperaturesByCity.csv') else 'GlobalLandTemperaturesByCity.csv'

print("Loading Kaggle urban temperature data (this might take a few seconds)...")
df_temp = pd.read_csv(file_path)

# Filter for the specific regional hub
city_name = "Bangalore"
print(f"Filtering data for {city_name}...")
df_city = df_temp[df_temp['City'] == city_name].copy()

# Clean dates and filter for years 1958 onward to match the CO2 baseline
df_city['dt'] = pd.to_datetime(df_city['dt'])
df_city = df_city.dropna(subset=['AverageTemperature'])
df_city = df_city[df_city['dt'].dt.year >= 1958]

# Calculate a 12-month rolling average to smooth out seasonal winter/summer extremes
df_city['Rolling_Temp'] = df_city['AverageTemperature'].rolling(window=12).mean()

# Plot the local urban temperature trend
plt.figure(figsize=(10, 5))
plt.plot(df_city['dt'], df_city['Rolling_Temp'], color='#ff7f0e', linewidth=1.5)
plt.title(f'Urban Climate Trend: {city_name} (1958 - Present)', fontsize=14)
plt.xlabel('Year', fontsize=12)
plt.ylabel('Average Temperature (°C)', fontsize=12)
plt.grid(True, linestyle='--', alpha=0.7)
plt.tight_layout()
plt.show()
```

```python
import pandas as pd
import matplotlib.pyplot as plt

# 1. Aggregate both datasets into annual averages for a cleaner comparison
df_co2_annual = df_co2.groupby(df_co2['date'].dt.year)['average'].mean().reset_index()
df_co2_annual.rename(columns={'date': 'Year', 'average': 'CO2'}, inplace=True)

df_city_annual = df_city.groupby(df_city['dt'].dt.year)['AverageTemperature'].mean().reset_index()
df_city_annual.rename(columns={'dt': 'Year', 'AverageTemperature': 'Temperature'}, inplace=True)

# 2. Merge the datasets precisely on the 'Year' column
df_merged = pd.merge(df_co2_annual, df_city_annual, on='Year')

# 3. Calculate the statistical correlation (Pearson)
correlation = df_merged['CO2'].corr(df_merged['Temperature'])
print(f"Statistical Correlation between Global CO2 and Bangalore Temperature: {correlation:.3f}")
print("(Note: A value closer to 1.0 indicates a strong positive correlation)\n")

# 4. Plot the final dual-axis overlay chart
fig, ax1 = plt.subplots(figsize=(10, 5))

# Axis 1: Local Temperature (Orange)
ax1.set_xlabel('Year', fontsize=12)
ax1.set_ylabel('Bangalore Temperature (°C)', color='#ff7f0e', fontsize=12, fontweight='bold')
ax1.plot(df_merged['Year'], df_merged['Temperature'], color='#ff7f0e', linewidth=2, label='Temperature')
ax1.tick_params(axis='y', labelcolor='#ff7f0e')
ax1.grid(True, linestyle='--', alpha=0.4)

# Axis 2: Global CO2 (Red)
ax2 = ax1.twinx()  # Create a second y-axis that shares the same x-axis
ax2.set_ylabel('Global Atmospheric CO2 (ppm)', color='#d62728', fontsize=12, fontweight='bold')
ax2.plot(df_merged['Year'], df_merged['CO2'], color='#d62728', linewidth=2, linestyle='-', label='CO2')
ax2.tick_params(axis='y', labelcolor='#d62728')

plt.title('Correlation: Global CO2 vs. Bangalore Urban Temperature', fontsize=14)
fig.tight_layout()
plt.savefig('correlation_chart.png')
plt.show()
```
